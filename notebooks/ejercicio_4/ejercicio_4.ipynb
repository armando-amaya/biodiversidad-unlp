```python
import os
import sys
from pathlib import Path

# Busca la carpeta 'code' de forma inteligente, sin asumir el cwd
current = Path(os.path.abspath(""))

if (current / "code").exists():
    code_path = current / "code"          # cwd es proyecto2026
elif current.name == "code":
    code_path = current                   # cwd es code/
elif current.parent.name == "code":
    code_path = current.parent            # cwd es code/notebooks/
elif current.parent.parent.name == "code":
    code_path = current.parent.parent     # cwd es code/notebooks/ejercicio_2/
else:
    raise FileNotFoundError("No se encontró la carpeta 'code'. Revisá desde dónde corre el notebook.")

#print("Directorio de trabajo:", os.getcwd())
#print("Carpeta 'code' detectada:", code_path)

if str(code_path) not in sys.path:
    sys.path.insert(0, str(code_path))

src_path = code_path / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

ruta_iadiza      = code_path / "raw_datasets" / "IADIZA" / "occurrence.txt"
ruta_inaturalist = code_path / "raw_datasets" / "iNaturalist" / "observations.csv"
ruta_xeno        = code_path / "raw_datasets" / "Xeno-canto" / "Occurrence.txt"
```
